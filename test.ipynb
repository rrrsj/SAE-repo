```python
import pandas as pd 
df = pd.read_parquet('/data/coding/sae/dataset/plain_text/partial-train/0000.parquet')
print(df['text'][0])
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
device = "cuda" # the device to load the model onto

model = AutoModelForCausalLM.from_pretrained(
    "./model_checkpoint",
    torch_dtype="auto",
    device_map="auto"
)
tokenizer = AutoTokenizer.from_pretrained("./model_checkpoint")
bos_token_id = tokenizer.bos_token_id  # BOS token ID（可能为 None）
eos_token_id = tokenizer.eos_token_id 
print(bos_token_id)
print(eos_token_id)
messages = "You are a helpful assistant."
text = tokenizer(
    messages,
)
print(text['input_ids'])
text = tokenizer.decode(text['input_ids'], skip_special_tokens=False)
print(text)
model_inputs = tokenizer([text], return_tensors="pt").to(device)
generated_ids = model(
    model_inputs.input_ids,
)
print(model_inputs.input_ids.shape)
print(generated_ids.logits.shape)

```
